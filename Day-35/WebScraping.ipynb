```python
import pandas as pd
import requests

from bs4 import BeautifulSoup
```

```python
# URL = "https://codegnan.com/"
URL = "https://timely-sunshine-e821b3.netlify.app/"
```

```python
page = requests.get(URL)
```

```python
page.status_code
```

```python
htmlCode = page.text
print(htmlCode)
```

```python
soup = BeautifulSoup(htmlCode, 'html.parser')
print(soup)
```

```python
content = soup.find_all('div', attrs={"class":"a"})
for i in content:
  print("Div:",i,end="\n\n\n\n")
```

```python
content = soup.find('div', attrs={'class':'name'})
text = content.text
print(text)
```

```python
content = soup.find('div', attrs={'class':'price'})
text = content.text
print(text)
```

```python
content = soup.find('div', attrs={'class':'image'})
text = content.text
print(text)
```

```python
img_tag=content.find('img')
print(img_tag)
```

```python
items=soup.find_all('div',class_='a')
names=[]
prices=[]
images=[]
for item in items:
  name=item.find('div',class_='name').text.strip()
  price=item.find('div',class_='price').text.strip()
  image_tag=item.find('div',class_='image').find('img')
  image_src=image_tag['src'] if image_tag else None

  names.append(name)
  prices.append(price)
  images.append(image_src)
print("Names:",names)
print("Prices:",prices)
print("Images:",images)
```

```python
df=pd.DataFrame({'Product_Name':names,'MRP':prices,'Image_SRC':images})
df
```

```python
df.to_csv('productdetails.csv',header=True,index=False)
```
